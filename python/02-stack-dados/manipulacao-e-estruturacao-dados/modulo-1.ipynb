## 📓 O que é um arquivo `.ipynb`?
Antes de começarmos, é importante entender a ferramenta que estamos usando. O formato `.ipynb` (Jupyter Notebook) é um ambiente de computação interativa que permite combinar três coisas em um único documento:

1. **Texto rico:** Explicações, fórmulas e documentação (como este texto que você lê agora).
2. **Código executável:** Blocos de código que podem ser rodados individualmente.
3. **Saída dinâmica:** O resultado do código (tabelas, gráficos, estatísticas) aparece logo abaixo da célula executada.

Ao contrário de um script `.py` comum, onde o programa corre do início ao fim, aqui você pode rodar, testar e visualizar dados parte por parte, o que facilita muito o aprendizado e a depuração.

## 💡 Células de Código vs. Células de Texto (Markdown)

No Jupyter Notebook cada bloco na tela é chamado de célula. Uma célula pode funcionar de duas maneiras principais e você precisa indicar ao sistema o que vai colocar nela:

* **Célula de Código (Code):** É o padrão do sistema. Serve para escrever e executar comandos em Python (como contas matemáticas ou instruções de programação). Se você escrever um texto comum aqui o sistema vai retornar um erro de sintaxe (*SyntaxError*).
* **Célula de Texto (Markdown):** Serve para escrever as suas explicações, anotações ou títulos.

### Como alterar o tipo da célula:
Se você quiser transformar uma célula de código em uma célula de texto para escrever uma anotação siga os passos abaixo:
1. Clique na célula que deseja modificar.
2. Olhe o menu superior do Jupyter Notebook e localize uma caixa de seleção que exibe a palavra **Code**.
3. Clique nessa caixa e mude a opção para **Markdown**.
4. Pressione as teclas **Shift + Enter** para salvar a formatação.

## 🐼 Introdução ao Pandas e Leitura de Dados
Agora que o nosso ambiente está pronto e sabemos usar as células, vamos conhecer a principal ferramenta de trabalho na área de dados: a biblioteca Pandas.

### 📦 Como importar e carregar o arquivo
Para usar ferramentas externas no Python, precisamos chamar elas no nosso código utilizando o comando `import`. Após a importação, o Pandas nos permite ler um arquivo CSV inteiro com um único comando.

💡 **Comparação de Código:** Vejam como abrir um arquivo exige muito controle manual de memória em C e como o Python simplifica esse processo para focarmos apenas na análise.

**Linguagem C (Leitura manual de arquivo):**
```c
#include <stdio.h>
#include <stdlib.h>

int main() {
    FILE *arquivo = fopen("dados_mineracao.csv", "r");
    char linha[2048];

    if (arquivo == NULL) {
        printf("Erro ao abrir o arquivo\n");
        return 1;
    }

    // Lendo e imprimindo apenas a primeira linha do cabecalho
    if (fgets(linha, sizeof(linha), arquivo)) {
        printf("%s", linha);
    }

    fclose(arquivo);
    return 0;
}

### ⚠️ Atenção ao Encoding (O problema dos acentos)

Se você tentar executar o comando básico de leitura em bases de dados brasileiras, é muito comum se deparar com um erro assustador chamado `UnicodeDecodeError`. 

Isso ocorre porque o padrão do Pandas para leitura de textos é o `utf-8`, mas muitos arquivos no Brasil (especialmente os do governo e do Excel) são salvos no padrão `latin1` ou `iso-8859-1` para suportar a nossa acentuação (ç, ã, é). 

Para corrigir isso, basta adicionar o parâmetro `encoding` no nosso comando de leitura, avisando ao Pandas qual é o formato correto do texto!

Veja como o código final deve ficar:

````python
import pandas as pd
tabela_mineracao = pd.read_csv("arquivos-e-imagens/producao_bruta_agencia_nacional_mineracao.csv", sep=",", encoding="latin1")

# Verifica as primeiras 5 linhas da nossa tabela
tabela_mineracao.head()
````

````python
# Verifica as ultimas 5 linhas da nossa tabela
tabela_mineracao.tail()
````

### 🏗️ Entendendo a Estrutura: DataFrames e Series
Quando usamos o comando `pd.read_csv`, o Pandas faz uma verdadeira mágica nos bastidores. Ele pega aquele texto puro do arquivo CSV e converte em uma estrutura de dados poderosa. A nossa variável `tabela_mineracao` virou exatamente isso: um **DataFrame**.

Mas o que isso significa na prática? No Pandas trabalhamos com dois conceitos estruturais fundamentais:

* **DataFrame:** É a tabela inteira bidimensional. Você pode imaginar o DataFrame como uma matriz completa ou a aba de uma planilha de Excel. Ele possui linhas e colunas. A variável `tabela_mineracao` inteira, cruzando os Estados e os valores de venda, é um DataFrame.
* **Series:** É uma única coluna isolada dessa tabela. Se você quiser separar apenas a coluna "UF" para analisar os Estados, o Pandas tratará essa coluna como uma Series (uma lista unidimensional). Em resumo, um DataFrame é um conjunto de várias Series coladas lado a lado!

💡 **Comparação de Estruturas de Dados:** Vejam a diferença entre definir essas estruturas manualmente em C e como o Python já entrega tudo pronto.

**Linguagem C (Simulando a estrutura com Structs e Vetores):**
```c
#include <stdio.h>

// Para simular a estrutura de um DataFrame em C,
// agrupamos os vetores (Series) dentro de uma unica estrutura
struct DataFrameMineracao {
    int ano_base[1000];
    char uf[1000][3];
    char substancia[1000][50];
    float series_valor_venda[1000]; // Aqui esta a nossa Series isolada!
};

int main() {
    // Instanciando a tabela inteira (o DataFrame)
    struct DataFrameMineracao tabela;

    // Acessando um valor especifico da nossa Series
    tabela.series_valor_venda[0] = 50000.50;

    return 0;
}

Em Python a estrutura fica muito mais simples, pois o gerenciamento de memória e a definição dos tipos de dados são feitos automaticamente. Não precisamos nos preocupar em definir o tamanho de vetores ou alocar espaço, o Pandas já resolve isso nos bastidores!

Podemos comprovar a estrutura da nossa variável principal executando a célula abaixo:

````python
# O comando type() revela qual e a classe da nossa variavel principal
type(tabela_mineracao)
````

````python
# O atributo .columns exibe os nomes exatos registrados na sua tabela
tabela_mineracao.columns
````
