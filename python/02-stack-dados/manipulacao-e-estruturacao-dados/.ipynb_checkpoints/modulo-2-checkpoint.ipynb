### O que vamos estruturar agora?
Vamos dominar quatro pilares fundamentais para organizar as nossas bases.
* **Filtragem de Linhas:** Isolando apenas as informações que queremos investigar.
* **Tratamento de Dados Nulos:** O que fazer com buracos e células vazias na nossa tabela.
* **Conversão de Tipos (Casting):** Ensinando o computador a diferenciar textos descritivos de números financeiros.
* **Limpeza de Colunas:** Descartando dados inúteis para economizar a memória da máquina.

## 🗂️ Conhecendo nosso Dataset

Neste módulo, vamos explorar os registros oficiais do **Centro de Investigação e Prevenção de Acidentes Aeronáuticos (CENIPA)**. O CENIPA é o órgão responsável por investigar acidentes e incidentes na aviação civil brasileira. Seu foco não é apontar culpados, mas sim identificar os fatores contribuintes para gerar recomendações de segurança e prevenir novas ocorrências no espaço aéreo nacional.

### 📊 O que tem na nossa base de dados?
A base que utilizaremos acompanha todas as ocorrências aeronáuticas notificadas em solo brasileiro desde 2007. Esse conjunto de dados é dividido em algumas tabelas (arquivos CSV) que se conectam. Ao carregar o arquivo principal de ocorrências, você encontrará diversas colunas detalhando os cenários, como:
* **Identificação e Classificação:** Códigos únicos do evento e se foi um incidente ou um acidente.
* **Localização e Data:** Cidade, estado (UF), país, aeródromo, coordenadas geográficas (latitude e longitude), dia e hora.
* **Investigação e Relatórios:** Status da investigação, se a aeronave foi liberada e detalhes sobre a publicação do relatório final.
* **Estatísticas do Evento:** Total de aeronaves envolvidas, total de recomendações geradas e se houve saída de pista.
* **Tabelas Complementares:** Além da tabela principal, a base oficial conta com arquivos extras focados nos Tipos de Ocorrência, Aeronaves, Fatores Contribuintes e Recomendações.

### 📥 Onde acessar os dados?
Para facilitar o andamento do laboratório, os arquivos já estão preparados e formatados para o nosso uso. Você tem duas opções para acessá-los:

1. **Usar o arquivo do repositório:** Se você fez o download ou clonou este projeto, os arquivos já estão no seu computador! Eles se encontram organizados dentro da pasta `arquivos-e-imagens` e se chama [CENIPA_ocorrencias.csv](arquivos-e-imagens/CENIPA_ocorrencias.csv)..
2. **Acessar a fonte oficial:** Caso queira entender como o Governo Federal publica esses dados para a sociedade, consultar o dicionário de variáveis ou baixar bases atualizadas, você pode acessar a página oficial no [Portal de Dados Abertos](https://dados.gov.br/dados/conjuntos-dados/ocorrencias-aeronauticas-da-aviacao-civil-brasileira).

```python
import pandas as pd

# Adicionamos o parametro encoding="latin1" para o Pandas entender os acentos em portugues
tabela_ocorrencias = pd.read_csv("arquivos-e-imagens/CENIPA_ocorrencias.csv", sep=";", encoding="latin1")

# Verifica as primeiras 5 linhas do nosso dataset
tabela_ocorrencias.head()
```

```python
# Exibe as ultimas 5 linhas do nosso dataset
tabela_ocorrencias.tail()
```

```python
# Filtro inicial
ocorrencias_sp = tabela_ocorrencias[tabela_ocorrencias["ocorrencia_uf"] == "SP"]
ocorrencias_sp.head()
```

```python
# Valores Nulos
print("Valores vazios por coluna:")
tabela_ocorrencias.isna().sum()
```

```python
# Casting
tabela_ocorrencias.dtypes
```

```python
colunas_apagadas = ["codigo_ocorrencia1", "codigo_ocorrencia2", "codigo_ocorrencia3", "codigo_ocorrencia4", "ocorrencia_pais"]
tabela_filtrada = tabela_ocorrencias.drop(colunas_apagadas, axis=1)
tabela_filtrada.head()
```

```python
nomes = {
    "ocorrencia_classificacao" : "ocorrencia_tipo",
    "ocorrencia_uf": "estado",
    "ocorrencia_cidade": "cidade"
}
tabela_filtrada = tabela_filtrada.rename(columns=nomes)
tabela_filtrada.head()
```
